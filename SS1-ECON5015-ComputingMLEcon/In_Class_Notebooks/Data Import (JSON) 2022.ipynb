# JSON

JSON is a format for storing/transferring data between applications.  It is a newer alternative to XML.

JSON has two basic data structures:  objects and arrays.

objects are like Python dictionaries, they are a set of key-value pairs.  The notation is similar to dictionaries in Python

    { California: Sacramento, Oregon: Salem, Washington: Olympia} is an object that maps state names to capitals.
    
arrays are like Python lists, they are an ordered set of values.  The notation is similar to lists in Python

    [California, Orgeon, Washington] is an array of state names.

The values in both objects and arrays can be strings, numbers, objects, arrays, true, false, or null.

The keys in objects must be strings.

# JSON to Python

Since all data in JSON has a corresponding data type in Python, importing JSON data is simple.

JSON to python parsers convert the JSON data into Python data structures.  All objects are converted to dictionaries.  All arrays are converted to lists. And strings/numbers/bools become corresponding Python data types.

The requests library includes a method for importing JSON data from the web (through HTTP).

The json library includes a function to convert a JSON string to python data structures.


# JSON Example

```text
{
  "Actors": [
    {
      "name": "Tom Cruise",
      "age": 56,
      "Born At": "Syracuse, NY",
      "Birthdate": "July 3, 1962",
      "married": false
    },
    {
      "name": "Robert Downey Jr.",
      "age": 53,
      "Born At": "New York City, NY",
      "Birthdate": "April 4, 1965",
      "married": true
    }
  ]
}
```

In the above example, the example itself is an object/dictionary with one key ("Actors") that maps to an array/list.  

The array/list contains two elements, each of which is an object/dictionary.

Those objects/dictionaries each contain five key-value pairs (name, age, Born At, Birthate, and married) which map to a string, number, string, string, bool.


````python
import json
f = open("json_example.txt", "r")
s = f.read()
s


x=json.loads(s)
x
type(x)
````

````python
type(x)
x["Actors"]
type(x["Actors"])
x["Actors"][0]
x["Actors"][0]["age"]
````

Typically, you will want to reorganize the data after importing to Python to put in a more useful format. 

For example, below I will put the above data into a DataFrame.

````python
import pandas as pd

Actors=pd.DataFrame()

for a in x["Actors"]:
    Actors=pd.concat([Actors,pd.DataFrame(a,index=pd.Index([0]))],ignore_index=True)

Actors
````

What if the JSON file is from the web, instead of on a local file.

You could use urllib to read in the file as a string and then use json the same as above.

A potentially easier way to do this is with the requests library, which includes its own JSON parser so no need to import json.

Example is below (note that output is the same as when using json.loads)

````python
import requests

url = 'http://ashafran.github.io/json_example.txt'
req = requests.get(url)
output = req.json()
output
````
