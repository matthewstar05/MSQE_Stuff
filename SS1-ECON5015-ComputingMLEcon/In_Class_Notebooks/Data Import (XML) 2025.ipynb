# Data Import (XML)

## XML

XML is a format for storing data to pass between applications.

XML files are made up of “tags” (which may have “attributes”).  The data provided in tags and attributes is application-specific.

XML files are a “text” format.  You can view them in a regular text editor (like Notepad on Windows) and see all the tags.



# Beautiful Soup

Beautiful soup is a package in python that provides functions to easily parse XML files and extract relevant information.

Note:  It can also be used to parse and extract information from HTML files.  This is useful for webscraping - i.e. importing data in Python from web pages.

We will import beautiful soup and also a package to help download/read XML files from the web.




````python
from bs4 import BeautifulSoup
import urllib.request as url
import pandas as pd
````

The BeautifulSoup class takes an XML file as input and creates an object with methods to parse the XML.

Conceptually, it is based on a hierarchical view of XML (example shown below).

Each tag has a parent tag, zero or more children tags, zero or more attributes, and possibly an untagged child (i.e. string text).

To extract all data into dataframe:

We iterate over every child of CATALOG (outer loop).  Each CD tag has a child for each column.  The column name is the tag name.   We create a dictionary that maps tag name to tag value for each child of the CD tag (the inner loop).  Then we append the dictionary to our data frame.

````python
xmldf = pd.DataFrame()
catalog=soup.contents[0]
for i in range(len(catalog.contents)):
    row=catalog.contents[i]
    cells = row.contents
    rowdict = {"id": int(row.attrs["id"])}
    for j in range(len(cells)):
        rowdict[cells[j].name] = cells[j].string
    xmldf=pd.concat([xmldf,pd.DataFrame(rowdict,index=pd.Index([0]))],ignore_index=True)
    
xmldf
````

In the above code, all data are strings (except for id which I coerced to int).  We could rewrite the code to coerce the price to float and the year to int.

````python
xmldf = pd.DataFrame()
catalog=soup.contents[0]
for i in range(len(catalog.contents)):
    row=catalog.contents[i]
    cells = row.contents
    rowdict = {"id": int(row.attrs["id"])}
    for j in range(len(cells)):
        if cells[j].name=="PRICE":
            rowdict[cells[j].name] = float(cells[j].string)
        elif cells[j].name=="YEAR":
            rowdict[cells[j].name] = int(cells[j].string)
        else:
            rowdict[cells[j].name] = cells[j].string
    xmldf=pd.concat([xmldf,pd.DataFrame(rowdict,index=pd.Index([0]))],ignore_index=True)
    
xmldf
````
