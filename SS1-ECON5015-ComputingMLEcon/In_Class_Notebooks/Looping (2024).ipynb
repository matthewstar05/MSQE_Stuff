It is very common when writing computer programs to need to do something repeatedly.

### range

The range data type is frequently used in conjunction with for loops, specifcally for "counter" loops.

### for loops

Example:



```python
l = []
for i in range(10): # range() is inclusive-exclusive
 l.append(i**2)   #body of loop

l
i
```


The indented lines after the for statement are the body of the loop.

In this example, there is only one line of code in the body of the loop.

The body is executed 10 times in this example.

Before executing the first time, there is an implicit assignment i = 0 (since 0 is the first value in the range object).  

Before the second execution, there is an implicit i=1, and so forth.  Each time, i is set equal to the next value in the range object.  At the end of the loop i is equal to 9, the last value in the range object.

## List Comprehension

The previous example creates a list with first 10 perfect squares (starting at 0).  An alternative way is to use list comprehensions.  The two approaches are basically the same except the syntax is a bit simpler for list comprehensions and list comprehensions should be slightly faster.  

List comprehensions are a useful alternative to a for loop for situations where the purpose of the for loop is to create a list with a set of elements according to some formula.  

```python
l2 = [i**2 for i in range(10)]
l2
```

You can add a conditional to a list comprehension.  In the following example, we create a list with perfect squares for only even integers.

```python
l3 = [i**2 for i in range(100) if (i%2==0 and i%9==1)]
l3
```

Although the “counter” loop from the previous examples is common (i.e. for i in range(10)), you can also loop over other iterables (like lists or strings) instead of a range object:

```python

for c in ["Albert","Bob","Carol"]:
    print(c)

for s in "Hello World":
    print(s)
```

In the first loops above, c is set equal to “Albert” the first time through the loop, c is set to “Bob” the second time, etc.

### Example

Suppose we want to find the minimum value (and its index) of a list l and we don’t know about the min function.

Algorithm:
1. Set our “working” minimum to be a very high number. (i.e. infinity)
2. For each element in list:
    * If less than working min, save it’s value as new working min and save index

Basic idea:  first element automatically becomes working min since any finite number < infinity.  

After that, only an element that is smaller than current working min will become new working min.

After all elements inspected, working min will be actual min.


```python

#l: list of  float or int values
l=[4,-5,100,0,-2,3]
currentmin = float("inf")
for i in range(len(l)):
    if l[i]<currentmin:
        currentmin = l[i]
        index = i
        
print(currentmin)
print(index)



```

### How do you decide whether to loop over indices (as in previous example) or loop over values?

Loop over indices if you need the index in what you're doing.  In the min example, we need to find both the min value and the min index so we loop over indices.

In next example below, we will find the minimum value in a list (just the minimum value, not the associated index).  Since we don't need the index, we can loop over values instread of over indices.

```python

currentmin = float("infinity")
for v in l:
    if v<currentmin:
        currentmin = v

print(currentmin)

```

## Algorithms

An algorithm is a programmatic approach to solving a specific problem that a programmer faces.

We often express algorithms in "pseudocode".  This means we don't use the exact syntax of any specific programming language but we write in a way such that other programmers could implement in the programming language of their choice.

Consider the problem of finding the integer square root of an integer (or determining if the square root is not an integer).  For example, for 25, we would find 5.  For 26, we would find that square root is not an integer.

One possible algorithm is called "exhaustive search" and the idea is try every possible answer until we get a "hit".  i.e. to start at 0 and try every integer between 0 and n, where n is the number we are finding the square root of.

Algorithm:

Set i to 0.

repeat until either i^2==n or i==n:

    i=i+1

After running through this algorithm, if i^2==n then i is the square root of n.  Otherwise, there is no integer square root of n.

Let's implement this in Python.

```python
n=14
i=0
while i**2!=n and i!=n:
    i=i+1
if i**2==n:
    print("The square root of "+str(n)+" is "+str(i)+".")
else:
    print("The square root of "+str(n)+" is not an integer.")
```

Next I am going to nest the code in a for loop so that we can run the algorithm for many values of n.  Specifically, we will try all integers from 0 to 29.

```python
for n in range(30):
    i=0
    while i**2!=n and i!=n:
        i=i+1
    if i**2==n:
        print("The square root of "+str(n)+" is "+str(i)+".")
    else:
        print("The square root of "+str(n)+" is not an integer.")
```

Exhaustive search doesn't work if there are an infinite number of possible solutions (or even if the number of possible solutions is finite but very very large).  

Grid search is a similar approach where you find an approximate solution by creating a finite grid of very close values and trying every point on the grid to get as close as possible to the correct solution.

For example, consider the problem of finding the square root of a positive integer, allowing for the non-integer solutions.  We can't try every possible solution (exhaustive search) because any positive real number is a possible solution and there are a lot of those!

Instead, we break the positive real numbers into points separated by a parameter $\epsilon$ where the smaller $\epsilon$ is, the more precise our solution will be (but also, the smaller $\epsilon$ is, the longer it will take our program to run).

If $\epsilon$ is 0.01, then we will try 0, 0.01, 0.02, 0.03, ..., n.  Our solution using grid search is the value $v$ that minimizes $|n-v^2|$.

Grid Search Algorithm:

Set e to small number (i.e. 0.01)

Set i to 0

Set working min to infinity

repeat until i>n:

    if |n-i*i| <working min:
    
        Set working min to |n-i*i|
        
        Set answer to i
  
    Set i to i+e

Now let's implement this algorithm in python.

```python
n=14
e=0.01
i=0
workingmin=float("inf")
while i<=n:
    error=abs(n-i**2)
    if error<workingmin:
        workingmin=error
        answer=i
    i=i+e

print("The square root of "+str(n)+" is "+str(answer)+".")
```

Again we can nest this code in a for loop to test with many values of n

```python
for n in range(30):
    e=0.01
    i=0
    workingmin=float("inf")
    while i<=n:
        error=abs(n-i**2)
        if error<workingmin:
            workingmin=error
            answer=i
        i=i+e
    print("The square root of "+str(n)+" is "+str(answer)+".")
```

## Bisection Search

Next we'll see an algorithm to find square root that is much faster than the previous one.  Here is the basic idea.  We know that the square root of x is between 0 and x (for $x\ge1$).  Since $y^2$ is an increasing function for y>0, we also know that if $y^2>x$, then $\sqrt{x}<y$ and if $y^2<x$ then $\sqrt{x}>y$   (for any $y$ such that $0 < y < x$).  

So, we make a guess somewhere in the middle of 0 and x.  If the guess is too high, we make a new guess in the middle of 0 and the first guess.  If the guess is too low, we make a new guess in the middle of the first guess and x.  Then we repeat until we are "close" to the right answer.

Algorithm:

Set epsilon to small number  (i.e. 0.001)

Set guess equal to x/2

Set delta equal to guess/2

Repeat until |guess^2-x| < epsilon

    if guess^2 > x then
    
        set guess = guess - delta
    
    else
    
        set guess = guess + delta
    Set delta = delta/2

```python
n=14
e=0.01
answer=n/2
delta=answer/2
while abs(answer**2-n)>e:
    if answer**2>n:
        answer=answer-delta
    else:
        answer=answer+delta
    delta=delta/2

print("The square root of "+str(n)+" is "+str(answer)+".")
```

Why is bisection search faster than grid search? 

There is less computation that needs to be done.  With grid search, we compute y^2 for every y value between 0 and n with our chosen precision.  For example, with precision=0.01 and n=14, we compute y^2 1400 times (100*14).  In contrast, with bisection search, we immediately rule out all y>7 in the first iteration (that's 700 values we don't need to try).  Then in the second iteration, we eliminate another 350 values.  Each iteration of the loop, we cut the problem in half.

Let's see how many iterations of the loop there are by adding a print statement.

```python
n=14
e=0.01
answer=n/2
delta=answer/2
while abs(answer**2-n)>e:
    print("Current guess: "+str(answer))
    if answer**2>n:
        answer=answer-delta
    else:
        answer=answer+delta
    delta=delta/2

print("The square root of "+str(n)+" is "+str(answer)+".")

```

So there are only 11 iterations of the loop compared to 1400 for grid search.  Much faster!  We can't really notice the difference because computers are so fast today but this would become noticeable if we want a more precise answer.

```python
import math
math.sqrt(14)
```

Suppose we want to find sqrt(14) to be accurate to 5 decimal places.  Now the difference is noticeable.  In fact, we can use bisection search to find accuracy to 10 decimal places instantly while grid search is slower with accuracy to just 5 decimal places.

```python
n=14
e=0.00000000001
answer=n/2
delta=answer/2
while abs(answer**2-n)>e:
    if answer**2>n:
        answer=answer-delta
    else:
        answer=answer+delta
    delta=delta/2

print("The square root of "+str(n)+" is "+str(answer)+".")
```

But here you have already seen the basic idea.  Algorithms are slower if they have more function evaluations.  And algorithms have more function evaluations if the number of iterations of loops increases.

So we are able to utilize a faster algorithm because we are taking advantage of something we know about the specific problem we are solving.  

## Numerical Optimization

A common problem in economics, econometrics, and machine learning is numerical optimization - i.e. finding the parameter values that yield the maximum or minimum of a function.

Examples:

Linear regression:  find parameters to minimize sum of squared errors

Maximum likelihood: find parameters to maximize likelihood of observing the data

The grid search approach can be utilized.  But, as we will see, it can be very slow if you have many parameters and/or you need a reasonable precise answer.   (Also, only works if parameters are bounded or if you have some other info about function).

An alternative approach is gradient-based search.  We will look specifically at Newton-Raphson, a well-known gradient-based method of optimization.

For all numerical optimization methods, one concern is that your solution is a local min/max instead of global min/max.  To alleviate this concern with gradient-based methods, you can try many different start values.  To alleviate this concern with grid search, you can try increasing precision. 


## Grid Search Example for Optimization

Example:

$f(x)=(x+2)^4$

$f'(x)=4(x+2)^3$

$f''(x)=12(x+2)^2$

We are trying to find the value of x that leads to minimum of f without solving analytically.  The correct answer is -2 (since I know f and can apply calculus), but we will proceed as if we do not know the answer  (for many realistic examples, we cannot analytically solves for minimum).

For grid search, the search space is bounded, so we will add the assumption that we are searching for a mininum on the interval $[-10,10]$.  If we want precision to nearest 0.01, then we can use the following code which checks every value from -10, -9.99, -9.98, ...., 9.99, 10.







```python
# Pre-defining f and its derivatives
def f(x):
    return (x+2)**4
def f1(x):
    return 4*(x+2)**3
def f2(x):
    return 12*(x+2)**2
```

```python
#Grid search
x=-10
p=0.01
workingmin=float("inf")
while x<=10:
    f_val=f(x)
    if f_val<workingmin:
        workingmin=f_val
        xmin=x
    x=x+p

print(xmin)
```

## Newton-Raphson

The basic idea is to make a starting guess.  Then, the next guess is the value that would be the min/max if the function was a quadratic.  Then, repeat until the first derivative is very close to 0 (i.e. you are close to a critical value for local min/max).  If the function was actually a quadratic, it would find the local/min on the first iteration.

Algorithm:

Set guess to initial guess

Repeat until $|f'(guess)|<\epsilon$:

    set guess to: old guess - f'(guess)/f''(guess)



```python
x=0  #initial guess
epsilon = 0.00001
while abs(f1(x))>epsilon:
    x = x-f1(x)/(f2(x))
print(x)
```

If f is a function of multiple variables, then the repeat condition is until the norm of the gradient of f is close to 0. 

Algorithm (multivariable)

Set guess to initial guess (a vector)

Repeat until $|\nabla(guess)|<\epsilon$:

   set guess to: old guess - $H^{-1}(guess)*\nabla(guess)$
