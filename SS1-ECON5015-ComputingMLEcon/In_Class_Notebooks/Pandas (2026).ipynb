### Series

Series is an object in Pandas that is an indexed, one-dimensional array.

The difference between a Series and the arrays from NumPy is that there is an explicit Index. 

The default Index, however, is the same as what we used to index arrays: a sequence of integers starting at 0.

In the example on the next slide, note that I am using the alias pd for pandas (for convenience).

```python
import numpy as np
import pandas as pd
s = pd.Series([1,3,5,7,9])
s
```

Compared to an array, note that there are two columns.  The first column is the Index column.  The second column is the actual data (i.e. the array).

In general, the explicit Index associated with a Series can complicate the syntax for selecting specific elements from a Series.

For now, though, since we are using the default Index, everything is roughly the same as with an array (except for the negative index case).

```python
s[0]
s[4]
s[1:3]
s[0:5:2]
s[:3]
```

```python
s[2:]
s[[1,3]]
s[[True,False,True,True,False]]
# s[-1]  does not work 
```

Now let's create some Series with other indexes.  In pd.Series below, the first argument is the data and the second argument is the index.

```python
uptolow = pd.Series(["a","b","c","d"],["A","B","C","D"])
uptolow
uptolow["C"]
uptolow["A":"C"]   #note that final index "C" is included
uptolow[["A","C"]]
```

What if a Series has integer indexes but not the default (implicit) indexes?

Some confusing things. If you provide a single index, the explicit index is used. If you provide a slice (using :) then implicit indexes are used.

```python
s = pd.Series([1,3,5,7,9],[1,2,3,4,5])
s
s[3]  #explicit index 3
s[1:3]  #implicit index 1 and 2 (3 excluded)
```

This is certainly very confusing, so it is best practice not to index Series in the ways shown above.

To alleviate this confusion, use loc[] to use explicit index and iloc[] to use implicit index.

By using loc and iloc, you clearly indicate which type of indexing you are intending to use, which makes your code much clearer.  I recommend using loc or iloc even in the case where the index is not integers (like the uptolow example above). 

```python
s[3]
s.loc[3]  #explicit index 3
s.iloc[3] #implicit index 3
s[1:3]
s.loc[1:3]  #explicit index 1 2 and 3
s.iloc[1:3] #implicit index 1 and 2
s[[2,4]]
s.loc[[2,4]]
s.iloc[[2,4]]
```

### DataFrames

DataFrames are like spreadsheets.  They have rows and (named) columns.  Each column is a Series.  Each Series (column) shares the same Index.

All elements in a column are of the same data type (since columns are a Series).

Different columns can have different data types.  Every column has the same length.

This format is helpful for standard data anaysis and econometrics.  You can think of the rows as the observations and the columns as the variables.  For example, if we have daily weather data for a year, then a DataFrame may have columns for hightemp, lowtemp, precip, and wind and 365 rows (one for each day in the year). 

In the example below, we will create a DataFrame with 2 columns:  the first 26 positive integers and the 26 lower case letters.


```python
num=np.arange(1,27,1)
num
let=np.array(list("abcdefghijklmnopqrstuvwxyz"))
let
df=pd.DataFrame({"numbers": num,"letters": let})
```

```python
df
```

Note that I created the data frame from two arrays, although I could have created num and let as Series instead of arrays.

Python automatically converts the array to a Series (using default Index), and if I reference one column of the data frame, it is of type Series.

You can reference a column of a data frame in two ways, shown below.

```python
type(df)
type(df.letters)

```

```python
df.letters
```

```python
df['letters']
```

```python
df.letters.iloc[0]
df.letters.iloc[10]
df.numbers.iloc[0]
df.numbers.iloc[25]
df.numbers.iloc[1:4]
df.letters.iloc[::2]
```

To select multiple rows of a data frame, we can index in the normal way.  

```python
df.iloc[1:11]  #selects rows 1-10
```

To select a single row from data frame, follow above example but select a single row.  This evaluates to a DataFrame with only one row.

If you want to select a single row as an array (rather than as a DataFrame), see next slide.






```python
df.iloc[2:3]  #second row as DataFrame

```

### Other attributes of DataFrames

You can think of a DataFrame as 2-dimensional array where the columns have names and the rows have names (or indices).  So, a DataFrame can be deconstructed into the values (the 2-dimensional array), the columns names, and the row indices.

The "values" attribute gets a representation of the data frame as a 2-dimensional array (omitting columns names and row indices).

A single entry in a 2-dimensional array can be indexed as [r,c] where r is the row and c the column. 

[r] yields the r-th row

[:,c] yields the c-th column

```python
type(df.values)
df.values[2,1]  #second row, first column
df.values[2]  #second row
df.values[:,1]  #first column
```

The "columns" attribute is of type Index and has all the column names.

You can reference specific elements of this Index (i.e. specific column names) in the same way as an array.

```python
df.columns
np.array(df.columns)  #change type to be array
df.columns[1]
```

The "index" attribute is of type Index and has all the row indices.

```python
df.index
df.index[2]
df.index[0:5]
np.array(df.index[0:5])
```

### Adding and removing a column to a DataFrame

```python
df["negative"]=np.arange(-1,-27,-1)
df

```

```python
df["mult"]=df.numbers*df.negative
df
```

```python
df=df.drop("mult",axis="columns")  #df.drop is creating a new dataframe which is then saved under the same variable name
df
```

### Adding and removing rows

To add rows, we will use the concat function in pandas.  We create a new dataframe with the new rows using pd.DataFrame and then concat the old dataframe and new dataframe.  

The concat function creates a new data frame (i.e. does not modify the input dataframes).  

Use ignore_index=True if using concat as shown below.

```python
df=pd.concat([df,pd.DataFrame({"numbers":[27],"letters": ["aa"],"negative": [-27]})],ignore_index=True)
df
```

```python
df=df.drop(3)
df
```

```python
df=df.reset_index(drop=True)
df
```
