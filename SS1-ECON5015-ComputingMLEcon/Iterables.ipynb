# Iterables
## GSE 524

Iterable objects in Python are objects which contain an ordered sequence of values, and those values can be accessed one at a time (i.e you can iterate over the values in an iterable object).  The values are referred to as "elements" of the iterable object.

There are a variety of types in Python that are iterable.  They have many things in common (which we will discuss here), and there are also important differences that we will discuss.

In the current slides, we will discuss the following types:  tuple, range, str, and list.  There are other iterable types (in Python as well as in extensions like NumPy) that we will disucss later.

Comparing iterable types:

* A str is a sequence of characters.  For example, "hello" is a string made up of 5 characters, "h", "e", "l", "l", and "o".
* A range is an ordered sequence of integers, where there is a start value, an end value, and a step.  For example range(0, 10, 3) is the sequence: 0, 3, 6, 9. 
* A tuple is a sequence of objects of any type.  So this is much more general than str or range.  It can be any conceivable sequence of objects.  A tuple is "immutable", which means once the object is created, it cannot be changed (i.e. cannot add objects to the sequence, or remove objects from the sequence, or change objects in the sequence).
* A list is just like a tuple but it is "mutable" which means you can modify a list by adding/removing/changing objects in the sequence.


# Strings

We start with strings, but the ways of accessing elements will apply to other iterables.

We can create a string by providing a literal string.  

The len function returns the number of elements (i.e. characters) in the string.

Sometimes we want to access one element (i.e. character) of a str (or a subset of elements/characters).

Iterables in python are indexed starting at 0. (i.e. first element is really the 0th element, and the element with index 1 is the second element of the list, etc.).  Therefore, the last element in an iterable has index n-1, where n is the length of the iterable.

```python
s = "Go Mustangs!"

len(s)
s[0]
s[1]
s[2]
s[11]
s[12]
```

# Slices

Slicing is a way to get a subset of an iterable.

```python
s[3:5]   #this means the 3rd element to the 5th element (not including the 5th), so elements 3 and 4
s[1:4]
s[:4]  #omitting number before colon implies 0:4
s[1:] #omitting number after colon implies end of iterable
s[-2] #means 2nd element from the end of iterable
```

```python
s[1:4:2]   #this means the 1st element to the 4th element (not including the 4th) and the 2 means to use a “step” of 2, so elements 1 then 3
s[::2]
s[::-1]
s[3::-1] 
s[3:1:-1]
s[::-2]
```

# Range

The range type is an ordered sequence of integers.  To create a range object, use the range function

range(start=0, stop, step=1)

With one argument, the argument is the stop value and the start is 0 and step is 1.

With two arguments, the first argument is start, second argument is stop, and step is 1.

The stop value is exclusive, meaning that the last value is the one before stop is reached (similar to slicing).

```python
r=range(3)
len(r)
r[0]
r[1]
r[2]
```

```python
r=range(1,6)
len(r)
r[0]
r[1]
r[2]
r[3]
r[4]
```

```python
r=range(10,0,-2)
len(r)
r[0]
r[1]
r[2]
r[3]
r[4]
r[1:4]
```

### Tuple

A tuple is a sequence of objects where each element can be of any type.

We can create a tuple by providing a comma-seperated sequence of objects in parentheses.  (If only element in the tuple, you should still provide a comma after the first element).

```python
t=(3,)
t
len(t)
t[0]
```

```python
t=("hello",4,3.2,False,7,(1,2,5))
t
len(t)
t[0]
t[1]
t[2]
t[3]
t[4]
t[5]
```

This tuple has 6 elements: a str, an int, a float, a bool, another int, and a tuple.

We call this a nested tuple, when one of the elements of a tuple is itself a tuple.

```python
len(t[5])
t[5][0]
```

```python
t[3:]
t[:5:2]
```

### Lists

Lists are just like tuples but they are mutable.  That means we an add elements, remove elements, and modify elements.  Lists are denoted with brackets instead of parentheses.

```python
l = [4,"calpoly",(3,True),-2] #list with 4 elements, an int, a str, a tuple and another int. 
len(l)
l[0]
l[1]
l[2]
l[3]
l[2][0]
l[2][1]
```

```python
l[1:3]
l[::-1]
```

```python
l.append("qwerty")  # adds element to the end
l
l.insert(1,(1,2,3,4)) # inserts tuple at index 1
l
l.remove(4)  #removes first matching element in the list
l
l.remove("-2")  #throws an error because no element matches "-2"
```

```python
l
l.remove(3)
```

```python
l
del l[0]  #use del to remove element at index 
#del finds element to remove by index whereas remove finds element to remove by value
l
```

```python
l
l[0]=5
l
l[2]="zxcvb"
l
```

### sum function

sum function takes iterable as argument and returns sum of elements (assuming elements are of type that can be summed, for example, range or list/tuple of numbers).

### concatenate

The + operator concatenates two lists/tuples/strings together, creating a new object.

### in

The in operator is preceded by a value and followed by an interable such as string/range/tuple/list.  It returns to True if the value is an element of the iterable and False otherwise

```python
sum(r)
sum((1,3,7,-3))
sum([1,2.3,True])

t+(1,3,7)
"hello"+"howdy"
[3,4,5]+[1,4,7,9]

```

```python
"h" in "hello"
"ell" in "hello"
"E" in "hello"
t
3.2 in t
2 in t
(1,2,5) in t
l
True in l
5 in l
"qwerty" in l
```

```python
len("")
len([])
len(())
len(range(0,0))
```
