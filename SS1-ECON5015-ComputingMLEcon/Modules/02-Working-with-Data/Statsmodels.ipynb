# A Very Brief Introduction to Statsmodels
## GSE 524

The statsmodels module provides functions to estimate a wide variety of statistical models, including OLS, logit, probit, times series, GMM, and a lot more.

https://www.statsmodels.org/

In these slides, I will show how to run a linear regression and view the results.



```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

df=pd.read_csv("data_1.csv")
df
```

```python
model=smf.ols("y ~ x1 + x2 + x3",data=df).fit()
model.summary()
```

```python
model.params
type(model.params)
model.params[0]
model.bse  #standard errors
type(model.bse)
```

```python
# measures of goodness of fit
model.rsquared
model.llf
model.aic
model.bic
```

```python
model.predict()  #predicted y values for actual data
model.predict(exog={"x1": 3,"x2":1,"x3":0})  #predicted y value for a new data point
```
