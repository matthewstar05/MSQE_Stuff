# Classes

Classes are a way for the programmer to define their own types.  

Classes can have "attributes", which are essentially variables associated with an instance of a class.

Classes can have "methods", which are essentially functions associated with an instance of a class.

If the variable x is an instance of some class and the class has attributes a and b, then you can reference those attributes as x.a and x.b.

Similarly, if the class has methods f and g, then you can call those methods x.f() and x.g().

We have already seen examples of using attributes and methods of classes at various points this quarter.

## How to define a class

You define a class using the keyword "class" and providing a name.  This doesn't create an object of the class, it just defines a class that you can later create an object for (similar to how def for functions doesn't call the function, it just defines a function that can be called later).

Indented and immediately following the class definition, you can define attributes and methods for the class.  In the example below, we define a class names Point that has two attributes, x and y, both of which will be initialized to value 0 whenever a Point object is created.  This example has no methods (see below for example with methods).

To create a Point object, we call the function Point().



```python
class Point:
    x=0
    y=0



p = Point()
p
type(p)
p.x
p.y

p.x = 1
p.y = 2
p.z = 3

p.x
p.y
p.z


```

## How to define methods

You can define methods for a class using "def" as part of the class definition, indenting on the lines after "class".

When defining a method, the first argument to the method should be "self" which is the way you reference the object itself from within the method.

However, when calling the method, you omit this argument; it will automatically define self to be the calling object.

You can optionally define a method for the class "\_\_init__".  This method will be automatically called every time you create a new object of this class.

In the example below, we use the \_\_init__ method to define the x and y attributes, which is preferable to the previous example.  By having default values of 0 for x and y in the method, we get the same default behavior as in the previous example but we offer the flexibility of providing other x and y values when creating the Point object.

```python
class Point:
    def __init__(self,x=0,y=0):
        self.x=x
        self.y=y

p = Point()
p
p.x
p.y

p2 = Point(3,4)
p2
p2.x
p2.y
```

```python
class Point:
    def __init__(self,x=0,y=0):
        self.x=x
        self.y=y
    
    def dist(self):
        import math
        return math.sqrt(self.x**2 + self.y**2)

p = Point()
p2 = Point(3,4)

p.dist()

p2.dist()

print(p)
```

You can optionally define the \_\_str__ method, which is called if you pass an instance of the class to the print function.  The \_\_str__ method creates a text representation of the class.

```python
class Point:
    def __init__(self,x=0,y=0):
        self.x=x
        self.y=y
        
    def __str__(self):
        return "("+str(self.x)+","+str(self.y)+")"
    
    def dist(self):
        import math
        return math.sqrt(self.x**2 + self.y**2)
    
p = Point()
p
print(p)
print(Point(4,3))
```

```python
class Point:
    def __init__(self,x=0,y=0):
        self.x=x
        self.y=y
        
    def __str__(self):
        return "("+str(self.x)+","+str(self.y)+")"
    
    def dist(self):
        import math
        return math.sqrt(self.x**2 + self.y**2)
    
    def __add__(self, other):
        return Point(self.x+other.x,self.y+other.y)
    
    def copy(self):
        import copy
        return copy.deepcopy(self)  #deepcopy copies the object and all objects the object references as attributes


def add_to_point(p):
    q=p.copy()
    q.x=q.x+1
    q.y=q.y+1
    return q

pnew = Point(6,7)

q=add_to_point(pnew)
q.x
q.y
pnew.x
pnew.y
```
