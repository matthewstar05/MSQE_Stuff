# Dictionaries (Hash Tables)
## Econ 5015


Motivating example:  Mapping keys to values

For example, maybe we have a set of students with names and students ID's and we want to lookup an ID given a name.  It will be helpful to have a mapping of names to ID's so that we can lookup the name and find the ID (quickly).

One solution to this example is to have two arrays, one of student names and one of student ID's (in the same order).  We can search the array until we find the name we are searching for.  For example, suppose it is the 16th name.  Then we can grab the 16th element in the ID array to get the array for that student.

The problem with this is that it can take a while to scan the name array to find the student in question, especially if there are a lot of students.

A better solution is dictionaries (Python data type) which are an implementation of a data structure called a hash table.

(A Series is also a good solution (as good as dictionary) but dictionary is a data type in Python without need for Pandas)



### dictionary

The dictionary data type in Python is a mapping of keys to values.  It is an implementation of an important data structure known as a "hash table".

keys can be of type int, float, str, or bool (and other types as well).  (but cannot be of type list or other unhashable type)

values can be of any type

When looking up a key to find the associated value, the key is provided to a hash function, which determines the value quickly (more on what I mean by quickly later).  



### Dictionary syntax



```python
d={}  #create empty dictionary
d

d["x"]=3  #create mapping from key "x" to value 3
d

d["y"]=6  #create mapping from key "y" to value 6
d

d["y"]  # lookup key "y"

d["z"]  # lookup key "z" which is not defined
```

```python
d.keys()  # all keys defined in d

"x" in d  #True if "x" is a key in d

"z" in d

d.values()  #all values in d

"x" in d.values()  # True if "x" is a value in d

6 in d.values()
```

```python
len(d)  #number of key-value pairs in d

d2 = {"a": 4, "b": "hello", "c": [1,2,6], 4: 3.14}  #a way to create a dictionary with mappings
d2
len(d2)
d2["c"]
d2[4]
```
