### Functions

We have already discussed many built-in functions like max, round, type, etc., plus the math functions in math module.

You can also write your own using the def keyword to define a new function.

Functions take inputs, called “arguments”, and have an output, the return value.

Functions have a name which is how you use the function.


### A simple example

```python
def absolutev(n):
    if (n<0):
        return (-n)
    else:
        return (n)

absolutev(-3)

```

### Calling a function

The body of a function (the indented code below the "def" line) is NOT executed when you define a function.  It is executed whenever you call a function.  In the example above, the function is called on the last line "absolutev(-3)".  If that line is removed, the body of the function is never executed.

When you call a function (for example, absolutev(-3) above), the variable n (the first argument) is set equal to -3, the value supplied.

Then, the sequence of code in the body of the function is executed in order from top to bottom until EITHER the end is reached, OR until a return value is provided using return.

If the code reaches a line with the keyword return, the function ends at that point and the value provided after return is the return value of the function.  No other lines of code in the body of the function are executed after return is reached.

If the end of the body of the function is reached prior to executing a statement with the return keyword, then the function does not have a return value.

### Calling a function (named arguments)

```python
import numpy as np
def myvariance(values, sample):
  l=len(values) 
  m = np.mean(values)
  if sample:  
    return np.sum((values-m)**2)/(l-1)
  else:
    return np.sum((values-m)**2)/l

myvariance(np.array([1,2,3]),True)
myvariance(sample=True,values=np.array([1,2,3]))

```

### Scope

An important concept related to functions is variable "scope" which roughly refers to “where” a variable is defined.

When you pass a value into a function, the argument defines a variable but only for use within that function.

For example, consider the function call:
absolutev(-3)

While executing the function, the variable n (the first argument of the function) is set equal to -3.  

When the function is complete, n is no longer defined.

### Global vs. Local Variables

Global variables:  when you define a variable at the console or in a script (but not within a function), it is a global variable.
* Global variables can be accessed from within functions as long as they have not been over-ridden.

Local variables: when you define a variable within a function (and also function arguments that are passed in), it is a local variable.
* Local variables can be accessed from within that function as well as any functions called by that function. 
* Once the function completes, local variables are no longer defined.

If a local variable defined in a function has the same name as a global variable, the locally defined value will be used within the function.  When the function completes, the globally defined value will persist.


### Example

```python
sample=False
myvariance(np.array([1,2,3]))
sample
```

Note that sample is a global variable equal to False.  Within myvariance, sample takes on the default value of the argument sample (True) which overrides the global defined value.  So, the result of the function call is 1, not 0.667.  When the function completes, the global value of sample (False) still holds.

What if an argument to a function is a mutable data type, like a list?

As described above, the argument name is a locally defined variable that ceases to be defined after the function completes.

However, since the object that is passed in is mutable, any changes to the object in the function will persist after the function completes.  



Note:  if you have studied other programming languages, you may recall "pass by value" and "pass by reference".  

"Pass by value" means the arguments to a function are copies of the values and changes do not persist when the function completes.  

"Pass by reference" means the arguments are pointers to a location in memory that stores the values - thus, changes to the values do persist when the function completes.  

In Python, essentially all immutable objects are passed by value.  Mutable objects are passed by reference.  

```python
def myfunction(l):
    r=l[::-1]
    return r

l=4
myfunction([1,2,3])
l
```

## An Example with Mutable Argument to a function

Let's write a function that takes a list as the only argument and returns a new list which has the first and last elements swapped.  The input list should not be modified.

Attempt 1:

```python
def f(l):
    temp=l[len(l)-1]
    l[len(l)-1]=l[0]
    l[0]=temp
    return l

l=[1,2,3,4]
f(l)
l
```

The return value is correct but it also modified the original list because the input list is mutable and so the changes that we made in the function affect the object even after function terminates.

Attempt 2:  create a local variable to modify

```python
def f2(l):
    r=l
    temp=r[len(r)-1]
    r[len(r)-1]=r[0]
    r[0]=temp
    return r

l=[1,2,3,4]
f2(l)
l
```

It still didn't work.  Why not?  Because the line "r=l" creates a local variable named r that refers to the same list object as l which is the input list.  Therefore, modifying r is modifying the input list.  So we didn't really address the problem at all with the r=l line.

So what do we do?  We need to explicitly create a new list.  In my correct solution below, I will use the copy method to create a new list with the same values as the input list.

```python
def f3(l):
    r=l.copy()
    temp=r[len(r)-1]
    r[len(r)-1]=r[0]
    r[0]=temp
    return r

l=[1,2,3,4]
f3(l)
l
    
```

Now consider the following function that takes a list as argument and returns a new list with the list order reversed.

Note that I didn't have to use copy because the first line is already creating a new object, NOT modifying the input list.


```python
def f4(l):
    l=l[::-1]
    return l

l=[1,2,3,4]
f4(l)
l
```
