### Mutability

All variables in Python represent an object.  An object can be of type int, str, float, list, other built-in types, or user-defined types.

Objects of some built-in types are immutable.  This means that, once created, they cannot be changed/modified.

Objects of other built-in types are mutable.  This means that the value(s) of the object can change.

### Immutable objects

The following built-in types are immutable:  int, bool, float, str (and some others like range, tuple)

When you create an integer object with the value 2, that object can never have another value.

Any attempt to modify the value of that object leads to a new object being created.

To illustrate this, we will use a few functions:

id() shows the location in memory where an object is stored

is:  a built-in function that checks if two expressions refer to the same object (similar but not the same as == which checks if two expressions have the same value)

```python
a=3
b=a
id(a)
id(b)
a is b
```

```python
a=2
id(a)
id(b)
a is b
```

The a=2 statement creates a new object with the value 2 and saves that as a.  The b object remains unchanged (with value 3).  You can see that the object id's are no longer the same and that "is" now yields False instead of True.

```python
c=5
d=c
id(c)
id(d)
c=c+1
id(c)
id(d)
c is d
```

Similar to the first example, the c=c+1 statment creates a new object with value 6 (which is what the expression c+1 evaluates to) and saves that as c.  The d object is unchanged with value 5.

### Mutable Objects

The following types are mutable:  list, dictionary, NumPy array, Pandas DataFrame/Series, user-defined objects

When you create a list with elements [1,3,5], that list object can be changed.  The values of any element can be changed without creating a new object and the list can have elements added or deleted without creating a new object.

If lists were immutable, then any time you changed/added a value, an entire new list would have to be created.  Therefore, mutability increases speed and reduces storage needs.  

Although mutability has the advantages described above, it can also create problems with your code if you don't understand all the consqeuences.  We will provide some examples to make this clear.   



```python
l = [1,3,5]
m=l
id(l)
id(m)
m is l
```

```python
l[0]=2  #Modifying l also modifies m since they refer to same object
l
m
id(l)
id(m)
m is l
```

```python
l.append(7) #Modifying l also modifies m since they refer to same object
l
m
id(l)
id(m)
m is l
```

The expression on the right side of the assignment statement (below) creates a new list object that contains the first 2 elements of the list l.

Since it creates a new object, which is then assigned to the variable l, the variable l now refers to a different object than above.

But the variable m was not changed, so it still refers to the original list.

```python
l=l[0:2] 
l
m
id(l)
id(m)
m is l
```

Sometimes you want to create a copy of a mutable object (such as a list) but as we just saw, a simple assignment will not work for mutable object.

Example:  revisting cumulative sum



```python
x=[1,2,3,4,5]

#create new list r that is cumulative sum of elements of x
r=x
for i in range(1,len(r)):
    r[i]=r[i]+r[i-1]

r
x
```

```python
x=[1,2,3,4,5]

#create new list r that is cumulative sum of elements of x
r=x.copy()
for i in range(1,len(r)):
    r[i]=r[i]+r[i-1]

r
x
```

Copy

The example above is for lists, but a copy method exists for many mutable objects.  For example:  arrays, Series, Dataframes, dictionaries

Copy vs Deep Copy

Note that copy creates a new list object but does not copy objects that are elements of the list.  To ensure you create a copy of all objects contained in a list, use deepcopy instead.

In the example below, I am importing the copy module which provides a deepcopy function.  For Series and Dataframes in Pandas, the copy method provides an option to do a deep copy.

```python
l=[1,2,[100,101,102],4]
c=l.copy()
l
c
```

```python
c[0]=-1
l
c
```

```python
c[2].append(103)
l
c
```

```python
import copy
l=[1,2,[100,101,102],4]
d=copy.deepcopy(l)
l
d
```

```python
d[2].append(103)
l
d
```

### Summary

If you create a variable that is equal to another variable, then both variables point to the same object.

If that object is mutable, then any modification of that object affects both variables.

If you use an assignment statement to re-define one of the variables to a new object, that does NOT affect the other variable, which will still point to the original object.



```python
a=2
b=2
a==b
a is b
c=[1,2,3]
d=[1,2,3]
c==d
c is d
```
