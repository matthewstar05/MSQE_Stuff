### Counting Operations

When designing an algorithm to solve a problem, the speed of execution will depend on several factors.
* The speed of the computer
* The number of basic operations
* File accesses?  Very slow.

Here I will focus on the number of operations when designing algorithms.


### The power function

Suppose we want to write a function to raise x to the nth power where n is a positive integer and both x and n are pre-defined.


```python
def power(x,n):
  r = 1  
  for i in range(n):
    r = r*x
  return r    

power(4,3)
```

Suppose an assignment operation takes some amount of time t.
    # assignment operations = 1+2n


Suppose an multiplication operation takes some amount of time u.
	# multiplication operations = n

Time to run power = (1+2n)\*t + n\*u = (2t+u)\*n + t = linear in n


### Big O Notation

Big O notation is concerned with speed as n gets large.  It provides an upper bound on the number of operations as a function of n.

If the number of operations is the sum of several terms (all functions of n), we drop all but the fastest growing term.  We also drop constant terms.

Example:
* Suppose one algorithm has $n^2+2n+5$ operations, another has $n^2+n+100$, and a third has $20n+200$.
* As n gets large, the fastest algorithm is the last one, and the slowest is first.  However, the difference in the first two is relatively small compared to the difference between those ones and the last one.
    1. $O(n^2)$
    2. $O(n^2)$
    3. $O(n)$
* Note that for n=2, the fastest is the first, slowest is 3rd (i.e. big O is about what happens when n gets large!)

Also, note that it is important to make sure you are using the correct “n” – the variable that can be scaled up to lead to large numbers of ops.  

For example, let’s say you are running one of the above algorithms approximately 1 million times, each time with an n between 5 and 10.

Then, you should use the first algorithm:
1. $10^2+2*10+5$=125 *1 million
2. $10^2+10+100$ = 210 * 1 million
3. $20*10+200$ = 400 * 1 million

The point here is that n is always small but the number of times the algorithm is run is what we need to be concerned with.
For large n, best to use 3rd algorithm.  For many repetitions with small n, best to use 1st.

#### The power function revisited

The power function is O(n).  i.e. the number of operations is proportional to n.

Can we do better?

Yes


```python
def power2(x,n):
  r = 1  
  while n>0:
    if (n % 2 > 0):
        r = r*x
    n = n // 2
    x = x*x
  return r

power2(4,3)
```

Each time through the loop, there are either 7 or 9 operations:  evaluate n>0, evaluate n % 2, evaluate (n % 2) > 0, evaluate n // 2, evaluate x\*x, and sometimes, evaluate r\*x   (plus 2 or 3 assignment operations)

How many times does this loop execute?


### The improved power function

Example, n=19

	Iteration 1: n=19
	Iteration 2: n=9
	Iteration 3: n=4
	Iteration 4: n=2
	Iteration 5: n=1

There is no 6th iteration because iteration 5 sets n=0.

Note: if n is a power of 2, there are exactly $log_2 (n) + 1$ iterations.  For other n, the number of iterations is $log_2(n)$ rounded up to the nearest integer.

### The improved power function

Since there are (worst case) 9 ops each time through the loop and the loop iterates (worst case) $log_2(n)+1$ times, the (worst case) total number of operations from looping is about $9 (log_2 (n)+1)$

Add to that the one operation at the beginning to set r=1 and the one at the end to return and we get:  $9(log_2(n) +1) +2$

(Note:  for simplicity, I am treating all operations as equivalent because in big-O notation, the differences don’t matter as long as they don’t depend on n).

Dropping the constants and slow growing terms, we get:

This algorithm is $O(log(n))$.

### Comparing the two power functions

log(n) grows much slower than n so the O(log(n)) algorithm will run much faster than the O(n) algorithm for large n.

n=1,000,000

power:  1,000,000 operations

power2: about 120 operations


```python
%timeit power(2,10000)
%timeit power2(2,10000)
```

```python
%timeit power(2,4)
%timeit power2(2,4)
```

### O vs. $\Theta$ vs. $\Omega$

Big O notation gives worst case speed of algorithm.

There are two other ways we can report speed of algorithm.

$\Omega$ gives best case speed

$\Theta$  is used if best and worst case are the same

For power2, the best case is also $\Omega$(log(n)) because the only difference is that there are 7 ops in the loop instead of 9 and we drop the constants anyway.

So, we can also say that power2 is $\Theta$(log(n)) since its best and worst case are functions of log(n)


### O vs. $\Theta$ vs. $\Omega$

An example with different best case and worst case running times:  search

Find the index of the first instance of a string in a list or return -1 if not found.



```python
def search(lis, s):
 n = len(lis)  
 for i in range(n):
    if (lis[i]==s):
        return i

 return -1

search([3,5,"hello","hllo"],"hello")
search([3,5,"hllo"],"hello")
```

Best case:  lis[0]==s, then loop only iterates once and the running time does not depend on the length of the vector

Worst case: s is not in lis, loop iterates n times

This search algorithm is $O(n)$ and $\Omega(1)$


### Optimizing Your Code

Big O Notation lets you compare the efficiency of algorithms while ignoring small implementation issues like an extra op here or there.  

Nonetheless, it is still beneficial to optimize code to reduce the number of ops inside loops.

For example, consider algorithm with $2n^2+2n+10$ ops.  If we could eliminate one op in the innermost loop, we may reduce it to $n^2+2n+10$ ops.

Both algorithms are $O(n^2)$ because we drop the constant term.  

However, the new algorithm could be about twice as fast.

Example:  n= 1000
* $1000^2+2*1000+5$ = 1002005
* $2*1000^2+2*1000+5$=2002005


### Optimizing Your Code

Optimization that affects the number of ops in slower growing terms don’t matter as much.

For example, consider algorithm with $2n^2+2n+10$ ops.  If we could eliminate one op in outer loop, we may reduce it to $2n^2+n+10$ ops.

As n gets large, the difference between these two algorithms is negligible.

Example:  n= 1000
* $2*1000^2+2*1000+5$ = 2002005
* $2*1000^2+1000+5$=2001005


### Optimizing Your Code

In general, don’t calculate the same value over and over again in a loop, move the calc to before the loop.  This won’t affect big-O since it just reduces the constant but it will speed up the code.

Note that a few extra calculations outside of loops never really matter for speed of execution.

The next slide has an example of how moving a calculation outside a loop improves performance.



```python
import numpy as np

def mymean(a):
    l=len(a)
    m=0
    for i in range(l):
       m=m+a[i]/l
    return m

def mymean2(a):
    l=len(a)
    m=0
    for i in range(l):
        m=m+a[i]
    return m/l

mymean([1,2,3,4])
mymean2([1,2,3,4])

np.random.seed(100)
data=np.random.random(10000)
%timeit mymean(data)
%timeit mymean2(data)
```

#### Optimizing Your Code

Now we will look at another example.

```python
def myvariance(values):
  s = 0  
  l=len(values)
  for i in range(l):
    s = s+(values[i]-np.mean(values))**2
  return (s/(l-1))

myvariance([1,2,3,4])
```

Note that np.mean(values) is calculated every time through the loop even though it’s value is unchanged.  Also, note that np.mean is a function that is $O(n)$ since it has to sum all n elements of values.

So, the function on the previous slide is $O(n^2)$.  The loop iterates n times and each time through, it calls a function that takes $O(n)$.

We can pull that function call out of the loop to save time every time through the loop.

The new version below is $O(n)$ because it calls an $O(n)$ function once and then it iterates a loop n times (with a constant number of ops in the body of the loop).  

In version 2b, I write the code to compute the mean myself rather than using the numpy function.  The point of 2b is to show that reducing O() efficiency of alogrithm is more important than using vectorized operations like numpy.mean.  i.e. 2b is still way faster than the original version, and only slightly slower than version 2.  

```python
def myvariance2(values):
  s = 0  
  l=len(values)
  m=np.mean(values)
  for i in range(l):
    s = s+(values[i]-m)**2
  return (s/(l-1))

def myvariance2b(values):
  s = 0  
  l=len(values)
  m=0
  for i in range(l):
     m=m+values[i]
  m = m/l
  for i in range(l):
    s = s+(values[i]-m)**2
  return (s/(l-1))

myvariance2([1,2,3,4])
myvariance2b([1,2,3,4])
```

```python
%timeit myvariance(data)
%timeit myvariance2(data)
%timeit myvariance2b(data)
```

### Optimizing your code

Final comment:  you can write myvariance without using a loop.  This is the best way to write variance in Python to take advantage of vectorization.   

In this final version, the O value will be the same as version 2 (O(n)), but you additionally get a significant speed improvement from using only vectorized ops and no loops.  

Note in the example below that the speed-up from full vectorization (with no loops) is even more significant that the speed-up from moving np.mean out of the loop.


```python
def myvariance3(values):
  l=len(values)
  m=np.mean(values)
  return np.sum((values-m)**2)/(l-1)


myvariance3([1,2,3,4])
```

```python
%timeit myvariance(data)
%timeit myvariance2(data)
%timeit myvariance3(data)
```
