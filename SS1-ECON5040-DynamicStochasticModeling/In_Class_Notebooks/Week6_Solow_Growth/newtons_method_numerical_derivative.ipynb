# Newton's Method with a Numerical Derivative

You supply only the function `f` and an initial guess `x0`.

The derivative is approximated numerically with the **central difference**

$$f'(x) \approx \frac{f(x+h)-f(x-h)}{2h}.$$

Newton's method then uses

$$x_{n+1}=x_n-\frac{f(x_n)}{f'(x_n)}.$$


```python
import numpy as np
```

## Numerical derivative


```python
def derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2*h)
```

## Newton's method


```python
def newton(f, x0, tol=1e-10, max_iter=100):
    x = x0

    for i in range(max_iter):
        if abs(f(x)) < tol:
            return x

        df = derivative(f, x)

        if abs(df) < 1e-12:
            raise ValueError("Derivative is too close to zero")

        x = x - f(x) / df

    raise ValueError("Newton's method did not converge")
```

## Enter the problem here

Change only `f` and `x0`.


```python
f = lambda x: np.sin(x/100)
x0 = 1.5
```

## Find the root


```python
root = newton(f, x0)

print("root =", root)
print("f(root) =", f(root))
```
