# The Bisection Method

We want to solve an equation $f(x)=0$.

If $f$ is continuous and $f(a)f(b)<0$, then a root lies between $a$ and $b$. Bisection repeatedly cuts this interval in half and keeps the half containing the sign change.


## Example

Solve $x^3-x-2=0$. Since $f(1)=-2$ and $f(2)=4$, a root lies in $(1,2)$.


```python
def f(x):
    return x**3 - x - 2

a, b = 1.0, 2.0
print('f(a) =', f(a))
print('f(b) =', f(b))
```

## One step

Compute the midpoint $m=(a+b)/2$. If $f(a)f(m)<0$, keep $[a,m]$; otherwise keep $[m,b]$.


```python
m = (a + b) / 2
print('midpoint =', m)
print('f(midpoint) =', f(m))
```

## The algorithm


```python
def bisection(f, a, b, tol=1e-8, max_iter=100):
    if f(a) * f(b) > 0:
        raise ValueError('The initial interval must bracket a root.')

    for iteration in range(max_iter):
        m = (a + b) / 2

        if abs(f(m)) < tol or (b - a) / 2 < tol:
            return m

        if f(a) * f(m) < 0:
            b = m
        else:
            a = m

    return (a + b) / 2
```

```python
root = bisection(f, 1.0, 2.0)
print('root =', root)
print('f(root) =', f(root))
```

## Watch the interval shrink

Each iteration halves the length of the interval.


```python
def bisection_table(f, a, b, n=10):
    print(f"{'iteration':>9} {'a':>12} {'b':>12} {'midpoint':>12} {'f(midpoint)':>14}")
    for iteration in range(1, n + 1):
        m = (a + b) / 2
        print(f'{iteration:9d} {a:12.8f} {b:12.8f} {m:12.8f} {f(m):14.8f}')
        if f(a) * f(m) < 0:
            b = m
        else:
            a = m

bisection_table(f, 1.0, 2.0)
```

## Convergence

If the original interval has length $b_0-a_0$, after $n$ iterations its length is

$$\frac{b_0-a_0}{2^n}.$$

Bisection is robust: for a continuous function, if the initial interval brackets a root, the method converges. Its drawback is that it uses no slope information, so it is generally slower than Newton's method.
