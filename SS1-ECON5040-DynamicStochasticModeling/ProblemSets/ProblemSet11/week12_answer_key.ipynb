## Problem 1 — A two-period insurance economy (1.25 points)

### (a) Arrow claims and the present-value budget

An $H$ claim pays one unit at $t=1$ only if state $H$ occurs; an $L$ claim pays one unit only in state $L$. Write their date-0 prices as $q_H,q_L$. If consumer $k$ purchases net positions $a_H^k,a_L^k$, then

$$c_0^k+q_Ha_H^k+q_La_L^k=y_0^k,$$
$$c_1^k(H)=y_1^k(H)+a_H^k,\qquad c_1^k(L)=y_1^k(L)+a_L^k.$$

Eliminating the asset positions yields

$$\boxed{c_0^k+q_Hc_1^k(H)+q_Lc_1^k(L)=y_0^k+q_Hy_1^k(H)+q_Ly_1^k(L).}$$

### (b) Arrow-pricing first-order conditions

The consumer maximizes $\ln c_0^k+\tfrac12\ln c_1^k(H)+\tfrac12\ln c_1^k(L)$ subject to the present-value constraint. The FOCs give

$$\frac{1}{c_0^k}=\lambda^k,\qquad \frac{1}{2c_1^k(s)}=\lambda^k q_s,$$

so

$$\boxed{q_s=\frac12\frac{c_0^k}{c_1^k(s)},\quad s\in\{H,L\}.}$$

### (c) Equilibrium

The agents have opposite date-1 endowments, while aggregate resources equal four units in every date/state. The symmetric candidate allocation is two units for each consumer in every date/state. The pricing FOCs then give $q_H=q_L=1/2$. Each consumer's present-value endowment is $2+(1/2)4=4$, and the proposed consumption bundle costs $2+(1/2)2+(1/2)2=4$. Thus the allocation is individually optimal, feasible, and an equilibrium.

### (d) Net positions

Agent A sells two $H$ claims and buys two $L$ claims: $a^A=(-2,2)$. Agent B takes $a^B=(2,-2)$. Each portfolio has zero initial cost, and both Arrow markets clear.

### (e) Interpretation

All risk here is **idiosyncratic**: when A receives four, B receives zero, and vice versa. Aggregate endowment is always four. Complete contingent claims allow agents to exchange these opposing exposures and each consume two in both states. If aggregate endowment varied across states, it would be impossible for *everyone* to consume the same amount in every state; complete markets would instead share aggregate risk efficiently.

```python
prob = np.array([0.5, 0.5])
y0 = np.array([2., 2.])
y1 = np.array([[4., 0.], [0., 4.]]) # agents A,B by states H,L
q = np.array([0.5, 0.5])
c0 = np.array([2., 2.])
c1 = np.full((2, 2), 2.)
a = c1 - y1
wealth = y0 + y1 @ q
cost = c0 + c1 @ q
print('Arrow prices:', q)
print('Consumption date 0:', c0, '; date 1 (rows A,B):\n', c1)
print('Net Arrow holdings (rows A,B):\n', a)
print('Wealth:', wealth, '; cost:', cost)
np.testing.assert_allclose(wealth, cost)
np.testing.assert_allclose(c0.sum(), y0.sum())
np.testing.assert_allclose(c1.sum(axis=0), y1.sum(axis=0))
np.testing.assert_allclose(a.sum(axis=0), 0)
np.testing.assert_allclose(c0[:,None] / (2*c1), np.tile(q, (2,1)))
print('All equilibrium checks passed.')
```

## Problem 2 — Constant aggregate endowment, infinite horizon (1.25 points)

### (a) Arrow prices and risk-free returns

Here $y^A=(0.8,0.2)'$, $y^B=(0.2,0.8)'$, so $y=(1,1)'$. The pricing equation is

$$Q_{ij}=\beta P_{ij}\left(\frac{y(j)}{y(i)}\right)^{-\gamma}.$$

Since aggregate endowment is constant, the consumption-growth factor is one and $Q=\beta P$. With $\beta=0.98$ and each transition probability $1/2$,

$$\boxed{Q=\begin{pmatrix}0.49&0.49\\0.49&0.49\end{pmatrix}.}$$

A risk-free claim is a portfolio of all next-state Arrow claims, so $R_i=1/\sum_jQ_{ij}=1/0.98\approx1.020408$ in either state.

### (b) Valuation operator and consumption shares

Because $P^2=P$, the geometric series simplifies to

$$V=(I-0.98P)^{-1}=I+\frac{0.98}{1-0.98}P=I+49P
=\begin{pmatrix}25.5&24.5\\24.5&25.5\end{pmatrix}.$$

The initial state is state 1 and financial wealth starts at zero. Hence

$$\alpha_A=\frac{[Vy^A]_1}{[Vy]_1}=\frac{25.3}{50}=0.506,\qquad \alpha_B=0.494.$$

The equilibrium consumption vectors are $c^A=(0.506,0.506)'$ and $c^B=(0.494,0.494)'$.

**Why not exactly one half?** Although their endowments are symmetric across states, the *initial state is fixed at state 1*. Agent A has a slightly higher present value of future endowment, which gives A a slightly larger consumption share.

### (c) Continuation financial wealth

The value of the difference between insured consumption and individual endowment is

$$\psi^k=V(\alpha_k y-y^k).$$

This gives $\psi^A=(0,0.6)'$ and $\psi^B=(0,-0.6)'$. The first entries are zero because the consumers start with zero financial wealth, and the two vectors add to zero because the securities are in zero net supply.

### (d) Interpretation

The consumers' endowments fluctuate across states, but total endowment does not. Complete markets eliminate the idiosyncratic fluctuations. Each consumer's equilibrium consumption is constant across states.

```python
P2 = np.array([[0.5, 0.5], [0.5, 0.5]])
beta2, gamma2 = 0.98, 2.0
yA2, yB2 = np.array([0.8, 0.2]), np.array([0.2, 0.8])
y2 = yA2 + yB2
s0 = 0
Q2 = beta2 * P2 * (y2[None,:] / y2[:,None])**(-gamma2)
R2 = 1 / Q2.sum(axis=1)
V2 = np.linalg.solve(np.eye(2)-Q2, np.eye(2))
alphaA2 = (V2 @ yA2)[s0] / (V2 @ y2)[s0]
alphaB2 = (V2 @ yB2)[s0] / (V2 @ y2)[s0]
cA2, cB2 = alphaA2*y2, alphaB2*y2
psiA2 = V2 @ (cA2-yA2)
psiB2 = V2 @ (cB2-yB2)
print('Q =\n', Q2, '\nR =', R2, '\nV =\n', V2)
print('shares:', alphaA2, alphaB2)
print('consumption A/B:', cA2, cB2)
print('financial wealth A/B:', psiA2, psiB2)
np.testing.assert_allclose(alphaA2+alphaB2, 1)
np.testing.assert_allclose(cA2+cB2, y2)
np.testing.assert_allclose(psiA2+psiB2, 0, atol=1e-10)
np.testing.assert_allclose([psiA2[s0],psiB2[s0]], 0, atol=1e-10)
np.testing.assert_allclose(cA2+Q2@psiA2, yA2+psiA2)
print('All equilibrium checks passed.')
```

## Problem 3 — Aggregate risk and Arrow prices (1.25 points)

### (a) Aggregate endowment and Arrow prices

The endowments are $y^A=(0.8,0.2)'$ and $y^B=(0.4,0.8)'$, so $y=(1.2,1.0)'$. Aggregate consumption is now lower in state 2. With $\beta=0.96$, $\gamma=2$ and

$$P=\begin{pmatrix}0.8&0.2\\0.3&0.7\end{pmatrix},$$

the pricing formula gives

$$Q=\begin{pmatrix}0.768&0.27648\\0.200&0.672\end{pmatrix}.$$

For example, $Q_{12}=0.96(0.2)(1/1.2)^{-2}=0.27648$. The state-2 payoff is valuable in state 1 because state 2 has lower aggregate consumption and thus higher marginal utility.

### (b) Risk-free returns

The gross risk-free return is the reciprocal of the row sum of $Q$. In state 1, the price of a sure next-period unit is $0.768+0.27648=1.04448$, so $R_1<1$. In state 2 the price is $0.2+0.672=0.872$, so $R_2>1$. The low return in state 1 reflects the high value of insurance against a transition to the low-consumption state.

### (c) Valuation and equilibrium shares

Compute $V=(I-Q)^{-1}$ and use the zero-initial-financial-wealth condition:

$$\alpha_k=\frac{[Vy^k]_1}{[Vy]_1},\qquad c^k(i)=\alpha_k y(i).$$

Both consumers' consumption moves with aggregate endowment, but their **shares** are state independent. The notebook calculates the exact numerical values rather than relying on rounded intermediate results.

### (d) Financial wealth and the recursive budget

Define $\psi^k=V(c^k-y^k)$. Since $(I-Q)V=I$,

$$(I-Q)\psi^k=c^k-y^k,$$

which rearranges to the statewise recursive budget identity

$$\boxed{c^k+Q\psi^k=y^k+\psi^k.}$$

This is a useful computational check. Also verify $\psi^k(1)=0$ and $\psi^A+\psi^B=0$.

### (e) Interpretation

Complete markets remove **individual** endowment risk conditional on aggregate resources: consumers obtain fixed shares of total consumption. They do **not** remove aggregate risk, because there is less total consumption available in state 2.

```python
P3 = np.array([[0.8, 0.2], [0.3, 0.7]])
beta3, gamma3 = 0.96, 2.0
yA3, yB3 = np.array([0.8, 0.2]), np.array([0.4, 0.8])
y3 = yA3 + yB3
Q3 = beta3 * P3 * (y3[None,:] / y3[:,None])**(-gamma3)
R3 = 1 / Q3.sum(axis=1)
V3 = np.linalg.solve(np.eye(2)-Q3, np.eye(2))
alphaA3 = (V3 @ yA3)[0] / (V3 @ y3)[0]
alphaB3 = (V3 @ yB3)[0] / (V3 @ y3)[0]
cA3, cB3 = alphaA3*y3, alphaB3*y3
psiA3, psiB3 = V3 @ (cA3-yA3), V3 @ (cB3-yB3)
print('Aggregate endowment:', y3)
print('Q =\n', Q3, '\nR =', R3, '\nV =\n', V3)
print('Shares A,B:', alphaA3, alphaB3)
print('Consumption A,B:', cA3, cB3)
print('Financial wealth A,B:', psiA3, psiB3)
np.testing.assert_allclose(cA3+cB3, y3)
np.testing.assert_allclose(cA3/y3, alphaA3)
np.testing.assert_allclose(cB3/y3, alphaB3)
np.testing.assert_allclose(psiA3+psiB3, 0, atol=1e-10)
np.testing.assert_allclose([psiA3[0], psiB3[0]], 0, atol=1e-10)
np.testing.assert_allclose(cA3+Q3@psiA3, yA3+psiA3)
np.testing.assert_allclose(cB3+Q3@psiB3, yB3+psiB3)
print('All equilibrium checks passed.')
```

### Visualizing the insurance result

The next figure uses **stacked bars** and **the same vertical scale in both panels**. The total height of each stack is aggregate endowment in that state. Before trading, the division of the stack depends on the individual endowment realization. After trading, the proportions of the two colored segments are constant across states. In this example, aggregate endowment is $1.2$ in state 1 and $1.0$ in state 2, so the total height changes: aggregate risk remains.

```python
states = np.arange(2)
labels = ['State 1', 'State 2']
fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharey=True)
for ax, a_vals, b_vals, title in [
    (axes[0], yA3, yB3, 'Before trade: endowments'),
    (axes[1], cA3, cB3, 'After trade: consumption')
]:
    ax.bar(states, a_vals, width=0.55, label='Consumer A')
    ax.bar(states, b_vals, width=0.55, bottom=a_vals, label='Consumer B')
    ax.set_xticks(states, labels)
    ax.set_ylim(0, 1.4)
    ax.set_title(title)
    ax.set_ylabel('Units of consumption')
axes[0].legend(loc='upper right')
plt.tight_layout()
plt.show()
```

## Problem 4 — Interpretation and proof (1.25 points)

### (a) Proving constant consumption shares

The household Euler equation for a one-period Arrow claim is

$$Q_{ij}=\beta P_{ij}\frac{u'(c^k(j))}{u'(c^k(i))}.$$

With common CRRA utility, $u'(c)=c^{-\gamma}$ and $\gamma>0$, so

$$Q_{ij}=\beta P_{ij}\left(\frac{c^k(j)}{c^k(i)}\right)^{-\gamma}.$$

All consumers face the same $Q_{ij}$ and $P_{ij}$. Equating the pricing formulas for A and B for any positive-probability transition yields

$$\frac{c^A(j)}{c^A(i)}=\frac{c^B(j)}{c^B(i)},$$

or

$$\boxed{\frac{c^A(j)}{c^B(j)}=\frac{c^A(i)}{c^B(i)}.}$$

If the state-transition graph is connected in the relevant sense, this ratio is the same in all states. Goods-market clearing then implies $c^k(i)=\alpha_k y(i)$ for constants $\alpha_k$ summing to one.

### (b) Arrow–Debreu versus sequential Arrow claims

An **Arrow–Debreu** arrangement trades at date 0 a claim for every possible future dated history. A **sequential Arrow** arrangement trades one-period state-contingent claims repeatedly as states unfold. Under complete markets, suitable solvency conditions, and consistent equilibrium prices, both arrangements can implement the same allocation. Sequential trade is more convenient for recursive computation.

### (c) When does the valuation operator exist?

The matrix geometric series

$$V=I+Q+Q^2+\cdots=(I-Q)^{-1}$$

converges if the **spectral radius** $\rho(Q)<1$. This ensures finite valuations for bounded endowment streams and makes the natural debt-limit and continuation-wealth formulas meaningful. It is stronger than merely writing a formal inverse: it establishes that the infinite discounted sum converges.

## Teaching takeaway

The central distinction is between **insurance against idiosyncratic endowment fluctuations**, which complete markets can accomplish, and **aggregate scarcity**, which no rearrangement of claims can eliminate. Common CRRA preferences turn efficient risk sharing into constant consumption shares, while the aggregate endowment process determines Arrow prices.
