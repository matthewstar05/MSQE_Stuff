```python
class Consumer:

    def __init__(self, w):
        "Initialize consumer with w dollars of wealth"
        self.wealth = w

    def earn(self, y):
        "The consumer earns y dollars"
        self.wealth += y

    def spend(self, x):
        "The consumer spends x dollars if feasible"
        new_wealth = self.wealth - x
        if new_wealth < 0:
            print("Insufficent funds")
        else:
            self.wealth = new_wealth
```

```python
# create a new instance of consumer with $100 in wealth
albert = Consumer(100)
```

```python
# add $900 of wealth using the earn method
albert.earn(900)
```

```python
# display the wealth
albert.wealth
```

```python
# reduce the wealth using the spend method
albert.spend(950)
```

```python
# directly access the data of the class and increase the wealth to $1,000,000
albert.wealth = 1000000
```

```python
# display the wealth as a formatted string
'$' + str(format(albert.wealth,',.2f'))
```

```python
# create another version of the class to implement encapsulation and abstraction
class Consumer2:
    "This is a simple class to model a consumer"
    def __init__(self, w):
        "Initialize consumer with w dollars of wealth"
        # wealth attribute is made private (encapsulation) through use of 2 leading underscores (called a 'dunder')
        self.__wealth = w
        
    # only way to (easily) change the wealth attribute is through these high-level methods (abstraction)
    def get_wealth(self):
        "Returns the current wealth of consumer"
        return self.__wealth

    def set_wealth(self, y):
        "The consumer increases wealth by y dollars"
        self.__wealth += y
```
